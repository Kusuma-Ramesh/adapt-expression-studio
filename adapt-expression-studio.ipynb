# **Project Summary -**

AdaptFER is a personalized facial expression recognition system developed using Deep Transfer Learning to improve emotion recognition accuracy for individual users. Conventional facial expression recognition models are trained on generalized datasets and often struggle to adapt to variations in facial features, lighting conditions, and personal expression styles. AdaptFER addresses this limitation by fine-tuning a pretrained DeepFER model using a small calibration dataset collected from the target user.

The system consists of a TensorFlow-based backend for model inference and a React-based frontend that provides an intuitive interface for calibration, evaluation, and real-time interaction. During calibration, the user records facial expressions corresponding to the seven standard emotions: Angry, Disgust, Fear, Happy, Sad, Surprise, and Neutral. These personalized samples are used to fine-tune the final classification layers while preserving the pretrained feature extraction capabilities of the backbone network.

Experimental evaluation demonstrates a significant improvement in personalized facial expression recognition. The generic DeepFER model achieved an accuracy of 57.14%, whereas the personalized AdaptFER model improved the accuracy to 84.76%, resulting in an absolute improvement of 27.62 percentage points and a relative error reduction of 64.44%. The personalized model also achieved higher confidence scores and improved recall across multiple emotion classes.

The project further includes an interactive dashboard for comparing model performance, monitoring calibration progress, visualizing evaluation metrics, and demonstrating expression-driven interaction scenarios. AdaptFER illustrates how lightweight personalization techniques can substantially improve facial expression recognition performance while maintaining computational efficiency and user-friendly deployment.

###Image Preprocessing Pipeline

Before training the facial expression recognition model, every facial image undergoes preprocessing to ensure consistency and improve model performance. Images are resized, converted to grayscale, normalized and formatted into tensors suitable for deep learning.

````python
import matplotlib.pyplot as plt
from PIL import Image
import numpy as np
import os
import random

dataset_base_path = "/content/archive-3/train"

sample_emotion = "neutral"

sample_folder = os.path.join(dataset_base_path, sample_emotion)

image_files = os.listdir(sample_folder)

random_image_filename = random.choice(image_files)

image_path = os.path.join(sample_folder, random_image_filename)

image = Image.open(image_path)

gray = image.convert("L")
resized = gray.resize((48,48))
normalized = np.array(resized)/255.0

fig,ax = plt.subplots(1,3,figsize=(12,4))

ax[0].imshow(image)
ax[0].set_title("Original")
ax[0].axis("off")

ax[1].imshow(resized,cmap="gray")
ax[1].set_title("48 × 48 Grayscale")
ax[1].axis("off")

ax[2].imshow(normalized,cmap="gray")
ax[2].set_title("Normalized")
ax[2].axis("off")

plt.tight_layout()
plt.show()
````

The preprocessing pipeline converts facial images into a standardized format suitable for model training. Grayscale conversion reduces computational complexity, resizing ensures a fixed input dimension of 48 × 48 pixels, and normalization scales pixel intensities between 0 and 1, leading to faster convergence and improved learning stability during training.
